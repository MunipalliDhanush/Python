```python
# From Feature_engine Importing the Winsorizer for cap extreme values(outliers) in dataset
from feature_engine.outliers import Winsorizer
```

```python
#It reduce the extreme Values for the Dataset
winsor=Winsorizer(capping_method='quantiles',tail='both',variables=list(num.columns))
#It will calculate cutoff thresholds from the data
winsor.fit(df)
```

```python
#- Stabilizes statistics , Keeps dataset size intact
df=winsor.transform(df)
```

```python
#Visulazing the data using heatmap for Mltivariate
sns.heatmap(df.corr(),annot=True)
```

```python
#Importing ScikitLearn for Model Selection and Model Training
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB,MultinomialNB
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

```python
#Determining the Dependent and Independent Column for Model Training
x=df.drop(['Seed_Variety'],axis=1)
y=df['Seed_Variety']
```

```python
#In this we can train the model and test. Need to be training is more and testing is less
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
#It is the  model_selection. which model we trainig and testing
model=GaussianNB()
model.fit(x_train,y_train)
```

```python
mnn=MultinomialNB()
mnn.fit(x_train,y_train)
```

```python

y1_pred=mnn.predict(x_test)
```

```python
accuracy_score(y_test,y1_pred)
```

```python
#Generates predictions from the trained model on unseen test data and stores them for evaluation.
y_pred=model.predict(x_test)
```

```python
accuracy_score(y_test,y_pred)
```

```python
print(classification_report(y_test,y_pred))
```

```python
print(classification_report(y_test,y1_pred))
```
